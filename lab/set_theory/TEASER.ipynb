##  To Belong ⊂ or to Exclude ⊄ 

Sets are lents to our human world view, a simplification of self **who am I** in relation to **What am I part of**

A set is a collection of objects consider as a single object. 
```
human organ = {cell𝖺, cell𝚋, ..., cell𝚗}
The school = {student𝖺, teacher𝖺, janitor𝖺, ...}
Diana's Sibblings = {Angelica, Andres, Joseph}
```

### Extensionality

It does not matter how we specify, count and organize the set memebers, but the existance of its members is what makes the set.

```
{a,a,b} = {a,b} = {b,a}
```
