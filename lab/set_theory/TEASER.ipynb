## ⊂ Belonging and ⊄ Exclusion

Sets are lenses to our human worldview, a simplification of **who am I** in relation to **What am I part of**

## Definition

A **set** is a defined collection of distinct, treated as a whole.

```python
human_organ = {"heart", "lung", "Liver", "Kidney"}
the_school = {"students", "teachers", "janitors", "administrators"}
Diana_siblings = {"Angelica", "Andres", "Joseph"}
```

### Extensionality
A set is defined *entirely by its members*, not by order or repetition.
$\{a, a, b\} = \{a, b\} = \{b, a\}$

This is why intersections return only **distinct** matches.

```python
from matplotlib_venn import venn2
import matplotlib.pyplot as plt

# Our sets (notice: duplicates in the real world resolve to distinct elements)
set1 = {1, 3, 4, 5, 6, 12}
set2 = {10, 4, 6, 2, 1, 7}

plt.figure(figsize=(7, 5))
venn = venn2(subsets=(set1, set2), set_labels=('Set A', 'Set B'))

# Styling for clarity
venn.get_patch_by_id('10').set_color('#FFB6C1')
venn.get_patch_by_id('01').set_color('#87CEEB')
venn.get_patch_by_id('11').set_color('#DDA0DD')

plt.title('Extensionality: {1,4,6} appear once, not twice', fontsize=12)
plt.show()

print(f"Set A       : {set1}")
print(f"Set B       : {set2}")
print(f"Intersection: {set1 & set2}")
```
